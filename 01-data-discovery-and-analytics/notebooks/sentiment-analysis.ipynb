## Gerando a base
Esta célula apenas simula o dado que será passado pelo Power BI ao script Python
O PBI (PowerQuery) cria um DATAFRAME PANDAS chamado dataset a partir do arquivo aberto, chamado dataset

```python
import pandas as pd

# Criar o DataFrame
dataset = pd.DataFrame({"Texto": [
    "Este produto é excelente, superou minhas expectativas!",
    "Achei muito ruim, não compraria novamente.",
    "O atendimento foi mediano, nada especial.",
    "Estou muito satisfeito com a qualidade.",
    "Estou nada satisfeito com a qualidade :-( ",
    "Terrível experiência, totalmente decepcionante!",
    "Estou muito pouco satisfeito com a qualidade."
    ]})
```

## Código principal

Este código será "colado" no Script PYTHON dentro do PowerQuery

```python
from transformers import pipeline

analisador = pipeline("sentiment-analysis", model="nlptown/bert-base-multilingual-uncased-sentiment", framework="pt") # Carregando o pipeline
resultados = analisador(dataset["Texto"].tolist()) # Extraindo os textos do Dataset e gerando notas
notas = []  # Criando lista para armazenar as notas

for resultado in resultados: # Percorrendo a lista com os resultados
    notas.append(int(resultado['label'].split()[0])) # Extraindo apenas o valor da nota de 1 a 5

dataset["Nota"] = notas # Adicionando as notas ao conjunto de dados
```

## Mostrando o resultado

```python
print(dataset)
```
